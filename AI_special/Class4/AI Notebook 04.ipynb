#### テキストの正規化

```python
# 改行と空白も除去

reg_text = re.sub('[\s\n]', '', reg_text)
print(reg_text)
```

#### 全記事の形態素解析

```python
# 全記事の名詞だけを収集

import os
import re
import codecs
import MeCab
import pandas as pd

dirs = ['it-life-hack', 'movie-enter']
docterm = []
label = []
focused_pos = '名詞'
tagger = MeCab.Tagger()

for i, d in enumerate(dirs):
    files = os.listdir('data_04/{0:s}/'.format(d))
    
    for file in files:
        text = ''
        for line in codecs.open('data_04/{0:s}/{1:s}'.format(d, file), 'r', 'utf-8'):
            text += line
        
        reg_text = re.sub('[0-9a-zA-Z]+', '', text)
        reg_text = re.sub('[:;/+\.-]', '', reg_text)
        reg_text = re.sub('[\n\s]', '', reg_text)
        
        result = tagger.parse(reg_text)
        
        tmp1 = []
        tmp2 = ''

        for elm in result.split('\n'):
            features = elm.split('\t')
            if len(features) > 1:
                info = features[4].split('-')
                pos = info[0]
                if pos == focused_pos:
                    tmp1.append(features[0])
        tmp2 = ' '.join(tmp1)
        docterm.append(tmp2)
        
        label.append(i)

pd.DataFrame(docterm).head()
```

#### 単語文書行列の作成

```python
# 文書×単語の行列を作成
# 面倒なのでscikit-learnのメソッドに依存

import numpy as np
from sklearn.feature_extraction.text import CountVectorizer

cv = CountVectorizer()
docterm_cv = cv.fit_transform(np.array(docterm))
docterm_cnt = docterm_cv.toarray()

pd.DataFrame(docterm_cnt).head()
```

```python
# 行列のサイズと単語のインデックス

print(pd.DataFrame(docterm_cnt).shape)
print(cv.get_feature_names_out()[:50])
```

```python
# 全記事における単語の出現数

wcnt = []

docterm_wcnt = np.sum(a=docterm_cnt, axis=0)
for w, cnt in zip(cv.get_feature_names_out(), docterm_wcnt):
    wcnt.append([w, cnt])
    
wcnt_df = pd.DataFrame(wcnt)
wcnt_df = wcnt_df.sort_values(1, ascending=False)
wcnt_df.head()
```

```python
# 極端に高/低頻度の単語はカット

cv = CountVectorizer(min_df=0.01, max_df=0.5) # 0.01以上0.5以下の単語を残す
docterm_cv = cv.fit_transform(np.array(docterm))
docterm_cnt = docterm_cv.toarray()

pd.DataFrame(docterm_cnt).head()
```

#### TF-IDFによる重み付け

ある単語のTF-IDF

$\log \frac {\rm \# all \; documents} {\rm \# documents \; containing \; a \; word} + 1$

特定の文書に頻出する場合に高い値を取る

```python
# TF-IDFの計算
# これもscikit-learnのメソッドに依存

from sklearn.feature_extraction.text import TfidfVectorizer

tv = TfidfVectorizer(min_df=0.01, max_df=0.5, sublinear_tf=True)
docterm_tv = tv.fit_transform(np.array(docterm))
docterm_tfidf = docterm_tv.toarray()

pd.DataFrame(docterm_tfidf).head()

# 品詞数が大分減っている気がする
```

#### LSTMによる記事分類モデル

```python
# データセットを作成
# 今回はとりあえず品詞を問わない

import os
import re
import codecs
import MeCab
import pandas as pd

dirs = ['it-life-hack', 'movie-enter']
wakati = []
label = []
tagger = MeCab.Tagger('-Owakati')

for i, d in enumerate(dirs):
    files = os.listdir('data_04/{0:s}/'.format(d))
    
    for file in files:
        text = ''
        for line in codecs.open('data_04/{0:s}/{1:s}'.format(d, file), 'r', 'utf-8'):
            text += line
        
        reg_text = re.sub('[0-9a-zA-Z]+', '', text)
        reg_text = re.sub('[:;/+\.-]', '', reg_text)
        reg_text = re.sub('[\n\s]', '', reg_text)
        
        result = tagger.parse(reg_text)
        wakati.append(result.split(' '))
        
        label.append(i)
        
print(len(wakati)) # 記事の数
print(wakati[0])
print(label[0])
```

#### 単語の数値化

```python
# 出現数の降順に単語を並べ替え

import itertools
from collections import Counter
import pandas as pd

word_freq = Counter(itertools.chain(*wakati))

dic = []
for word_uniq in word_freq.most_common():
    dic.append(word_uniq[0])
    
print(len(dic))
pd.DataFrame(dic).head()

# 役に立たなそう
```

```python
# 番号をつけてdictionary変数に格納

dic_inv = {}
for i, word_uniq in enumerate(dic, start=1):
    dic_inv.update({word_uniq: i})
    
print(len(dic_inv))
```

```python
# wakatiに含まれるすべての単語をIDに変換

wakati_id = [[dic_inv[word] for word in waka] for waka in wakati]

print(len(wakati_id))
print(wakati_id[0])
```

```python
# 文書によってIDリストの長さがバラバラ
# リストの長さを整える

from keras.utils import pad_sequences
import numpy as np

wakati_id = pad_sequences(np.array(wakati_id, dtype=object), padding='post')
label = np.array(label)

print(wakati_id[0])
```

#### LSTMのモデル

embedding $\rightarrow$ LSTM $\rightarrow$ dense

```python
# LSTMの構築

from keras.models import Sequential
from keras.layers import Embedding, LSTM, Dense

model = Sequential()
model.add(Embedding(14885, 512, input_shape=(399, )))
model.add(LSTM(128, dropout=0.5))
model.add(Dense(1, activation='sigmoid'))

model.summary()
```

```python
# 学習の実行

model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
hist = model.fit(wakati_id, label, batch_size=32, verbose=1, epochs=5, validation_split=0.2) # 20%を検証データに

# 参考書にはepochs=30と書いてあったがとても終わりそうにない
```

```python
# 誤差の履歴

import matplotlib.pyplot as plt

plt.plot(hist.history['loss'])
plt.plot(hist.history['val_loss'])
plt.ylabel('loss')
plt.xlabel('epoch')
plt.legend(['train', 'val'], loc='upper right')
plt.show()
```

```python
# 精度の履歴

plt.plot(hist.history['accuracy'])
plt.plot(hist.history['val_accuracy'])
plt.ylabel('accuracy')
plt.xlabel('epoch')
plt.legend(['train', 'val'], loc='upper right')
plt.show()

# 本当は25epochくらいからやっと精度が良くなるらしい
```

#### トピック抽出

```python
# まずはデータセット作成

import os
import re
import codecs
import MeCab
import pandas as pd

dirs = ['it-life-hack', 'movie-enter']
docterm = []
label = []
focused_pos = '名詞'
tagger = MeCab.Tagger()

for i, d in enumerate(dirs):
    files = os.listdir('data_04/{0:s}/'.format(d))
    
    for file in files:
        text = ''
        for line in codecs.open('data_04/{0:s}/{1:s}'.format(d, file), 'r', 'utf-8'):
            text += line
        
        reg_text = re.sub('[0-9a-zA-Z]+', '', text)
        reg_text = re.sub('[:;/+\.-]', '', reg_text)
        reg_text = re.sub('[\n\s]', '', reg_text)
        
        result = tagger.parse(reg_text)
        
        tmp1 = []
        tmp2 = ''

        for elm in result.split('\n'):
            features = elm.split('\t')
            if len(features) > 1:
                info = features[4].split('-')
                pos = info[0]
                if pos == focused_pos:
                    tmp1.append(features[0])
        tmp2 = ' '.join(tmp1)
        docterm.append(tmp2)
        
        label.append(i)

pd.DataFrame(docterm).head()
```

```python
# TF-IDFの計算

from sklearn.feature_extraction.text import TfidfVectorizer

tv = TfidfVectorizer(min_df=0.01, max_df=0.5, sublinear_tf=True)
docterm_tv = tv.fit_transform(np.array(docterm))
docterm_tfidf = docterm_tv.toarray()

pd.DataFrame(docterm_tfidf).head()
```

```python
# 最後にlabel列を結合

label = pd.DataFrame(label)
label = label.rename(columns={0: 'label'})

docterm_df = pd.concat([pd.DataFrame(docterm_tfidf), label], axis=1)
docterm_df.head()
```

#### 類似度の計算

$\cos$類似度を使うのが一般的

```python
# cos類似度の計算
# これもscikit-learnのメソッドに依存

from sklearn.metrics.pairwise import cosine_similarity

docterm_0 = docterm_df[docterm_df['label']==0] # it-life-hackの方
docterm_0 = docterm_0.drop('label', axis=1)

sim0 = cosine_similarity(docterm_0.T)
sim0_df = pd.DataFrame(sim0)
sim0_df.head()

# 対角成分は1
```

#### 共起語エッジリストの作成

```python
# 2つの単語のペアと類似度をまとめる

sim0_stack = sim0_df.stack()

index = pd.Series(sim0_stack.index.values)
value = pd.Series(sim0_stack.values)

index.head()
```

```python
value.head()
```

```python
# 共起語エッジリストの作成

tmp3 = []
tmp4 = []

for i in range(len(index)):
    if value[i] >= 0.5 and value[i] <= 0.9:
        tmp1 = str(index[i][0]) + ' ' + str(index[i][1])
        tmp2 = [int(s) for s in tmp1.split()]
        tmp3.append(tmp2)
        tmp4 = np.append(tmp4, value[i])
        
tmp3 = pd.DataFrame(tmp3)
tmp3 = tmp3.rename(columns={0: 'node1', 1: 'node2'})
tmp4 = pd.DataFrame(tmp4)
tmp4 = tmp4.rename(columns={0: 'weight'})
sim0_list = pd.concat([tmp3, tmp4], axis=1)

sim0_list.head()
```

```python
# movie-enterでも同様の作業

docterm_1 = docterm_df[docterm_df['label']==1]
docterm_1 = docterm_1.drop('label', axis=1)

sim1 = cosine_similarity(docterm_1.T)
sim1_df = pd.DataFrame(sim1)

sim1_stack = sim1_df.stack()

index = pd.Series(sim1_stack.index.values)
value = pd.Series(sim1_stack.values)

tmp3 = []
tmp4 = []

for i in range(len(index)):
    if value[i] >= 0.5 and value[i] <= 0.9:
        tmp1 = str(index[i][0]) + ' ' + str(index[i][1])
        tmp2 = [int(s) for s in tmp1.split()]
        tmp3.append(tmp2)
        tmp4 = np.append(tmp4, value[i])
        
tmp3 = pd.DataFrame(tmp3)
tmp3 = tmp3.rename(columns={0: 'node1', 1: 'node2'})
tmp4 = pd.DataFrame(tmp4)
tmp4 = tmp4.rename(columns={0: 'weight'})
sim1_list = pd.concat([tmp3, tmp4], axis=1)

sim1_list.head()
```

#### 共起語ネットワークの作成

```python
# networkxライブラリを用いた共起語ネットワークの作成

import networkx as nx

G_corlist = nx.from_pandas_edgelist(sim1_list, 'node1', 'node2', ['weight'], nx.Graph)
print(G_corlist.nodes())
print(G_corlist.edges(data=True))
```

```python
# ネットワークの描画

import matplotlib.pyplot as plt

plt.figure(figsize=(10, 10))
pos = nx.spring_layout(G_corlist) # springモデル
nx.draw_networkx(G_corlist, pos)
plt.axis('off')
plt.show()
```

#### ネットワークの特徴

```python
# クラスタ係数

print(nx.average_clustering(G_corlist))
```

```python
# 媒介中心性

bc = nx.betweenness_centrality(G_corlist)
for k, v in sorted(bc.items(), key=lambda x: -x[1]):
    print(str(k) + ': ' + str(v))
```

#### 共起語コミュニティの抽出

```python
# モジュラリティ(modularity)の計算

from networkx.algorithms.community import greedy_modularity_communities

cm_corlist = list(greedy_modularity_communities(G_corlist))
print(cm_corlist)
```

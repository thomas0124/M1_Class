```python
# yfinanceを用いたトヨタとTOPIXの株価データ取得
# 期間は5年分

import datetime
import pandas as pd
import numpy as np
import yfinance as yf
import matplotlib.pyplot as plt

target = "7203.T"
symbols = ("1306.T", target)

data = yf.download(symbols, period='5y', interval = "1d")
print(data)
```

```python
# データを整形

df = data['Close']
df.index = df.index.rename('Date')
df.dropna()
df.head(10)
```
